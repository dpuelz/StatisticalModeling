```python
import torch
import torchvision
import torchvision.transforms as transforms
import torch.nn as nn
import torch.nn.functional as F
import matplotlib.pyplot as plt
import numpy as np
import torch.optim as optim
import tqdm
```

```python

# Download the MNIST dataset and create data loaders
transform = transforms.Compose([transforms.ToTensor(),
                                transforms.Normalize((0.5,), (0.5,))])

trainset = torchvision.datasets.MNIST(root='./data', train=True,
                                      download=True, transform=transform)
trainloader = torch.utils.data.DataLoader(trainset, batch_size=64,
                                          shuffle=True, num_workers=2)

testset = torchvision.datasets.MNIST(root='./data', train=False,
                                     download=True, transform=transform)
testloader = torch.utils.data.DataLoader(testset, batch_size=64,
                                         shuffle=False, num_workers=2)

```

```python
# Function to display a sample of images
def imshow(img):
    img = img / 2 + 0.5
    npimg = img.numpy()
    plt.imshow(np.transpose(npimg, (1, 2, 0)))
    plt.show()

# Get a batch of training images and display them
dataiter = iter(trainloader)
images, labels = dataiter.next()

imshow(torchvision.utils.make_grid(images[:16]))
print('Labels:', ' '.join('%d' % labels[j] for j in range(16)))
```

```python
# Let's define a simple CNN to classify MNIST digits (0–9)

class Net(nn.Module):
    def __init__(self):
        super(Net, self).__init__()
        
        # First convolution layer:
        # Input:  (batch_size, 1, 28, 28)
        # 1 input channel (grayscale), 32 output channels
        # 3x3 kernel, stride 1
        # Output spatial size: (28 - 3)/1 + 1 = 26
        # Output: (batch_size, 32, 26, 26)
        self.conv1 = nn.Conv2d(1, 32, 3, 1)

        # Second convolution layer:
        # Input:  (batch_size, 32, 26, 26)
        # 32 input channels → 64 output channels
        # 3x3 kernel, stride 1
        # Output spatial size: (26 - 3)/1 + 1 = 24
        # Output: (batch_size, 64, 24, 24)
        self.conv2 = nn.Conv2d(32, 64, 3, 1)

        # After max pooling (2x2), spatial dims go from 24x24 → 12x12
        # So we have 64 feature maps of size 12x12
        # 64 * 12 * 12 = 9216 features total
        # This is why the first linear layer has 9216 inputs
        self.fc1 = nn.Linear(9216, 128)

        # Final fully connected layer:
        # 128 hidden units → 10 output classes (digits 0–9)
        self.fc2 = nn.Linear(128, 10)

    def forward(self, x):
        # x shape: (batch_size, 1, 28, 28)

        x = self.conv1(x)          # → (batch_size, 32, 26, 26)
        x = F.relu(x)              # Apply nonlinearity

        x = self.conv2(x)          # → (batch_size, 64, 24, 24)
        x = F.relu(x)              # Apply nonlinearity

        x = F.max_pool2d(x, 2)     # 2x2 max pooling → (batch_size, 64, 12, 12)

        x = torch.flatten(x, 1)    # Flatten all but batch dimension
                                   # → (batch_size, 9216)

        x = self.fc1(x)            # → (batch_size, 128)
        x = F.relu(x)

        x = self.fc2(x)            # → (batch_size, 10)

        # Convert raw scores to log-probabilities
        return F.log_softmax(x, dim=1)

net = Net()
print(net)
```

```python
# Set the device to GPU if available
device = torch.device("cuda:0" if torch.cuda.is_available() else "cpu")
net.to(device)

# Define the loss function and the optimizer
criterion = nn.CrossEntropyLoss()
optimizer = optim.Adam(net.parameters())

```

```python
# Number of training epochs
num_epochs = 5

testiter = iter(testloader)
# Main training loop
for epoch in range(num_epochs):
    running_loss = 0.0
    for i, data in enumerate(trainloader, 0):
        inputs, labels = data
        inputs, labels = inputs.to(device), labels.to(device)

        # Zero the parameter gradients
        optimizer.zero_grad()

        # Forward, backward, and optimize
        outputs = net(inputs)
        loss = criterion(outputs, labels)
        loss.backward()
        optimizer.step()

        running_loss += loss.item()

        if i % 20 == 0:
            with torch.no_grad():
                # Show a sample of images from the test data and model's predictions
                test_images, test_labels = testiter.next()
                test_images, test_labels = test_images.to(device), test_labels.to(device)

                imshow(torchvision.utils.make_grid(test_images[:8].cpu()))
                test_outputs = net(test_images[:8])
                _, predicted = torch.max(test_outputs, 1)
                print('Predicted:', ' '.join('%d' % predicted[j] for j in range(8)))

    # Print average loss for the epoch
    print(f"Epoch {epoch+1}, Loss: {running_loss / (i+1)}")


print('Finished Training')
```
